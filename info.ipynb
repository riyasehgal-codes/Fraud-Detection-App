## FRAUD DETECTION APP 

DATASET : AIML Dataset.csv ( https://www.kaggle.com/datasets/amanalisiddiqui/fraud-detection-dataset?resource=download )

app.py : streamlit app to create a webpage that allows users to detect Fraud based on transactionType, amount, oldbalanceOrg, newbalanceOrig,oldbalanceDest,newbalanceDest

main.ipynb : used for data analysis, data cleaning, data visualization, EDA, feature engineering, model training and testing

fraud_detection_pipeline.pkl : trained model and pipeline
